# E117: on the 67M, does VPD's switched-on set say more than the same number of neurons?

## The answer

Partly. VPD's switched-on set repeats across two seeds more for copy maps than for random maps, and the same number of
neurons does not. Neither set carries much of the fine-tune's gain, and neither tells the maps apart on what it
carries. At full width VPD's re-switching share puts copy maps and lowercase well above random maps and ciphers; the
neuron share reverses on the maps. On lowercase and the ciphers VPD switches on nothing at all, so those panels are
unreadable.


## What we were trying to test

The user asked for VPD's readings and the same readings on neurons side by side, on states where VPD has a reading.
The states are the 67M's copy maps and random maps from E12 and E46, two tasks each, and lowercase and two ciphers from
E53, two seeds each. Copy maps and lowercase were built to use something the model can already do; random maps and
ciphers were built not to. Those classes are assumptions from the task design.

- **Switched-on set.** VPD: alive subcomponents whose mean causal importance at the read position rises by more than
  0.3 over the unchanged model's. Neurons: the same number of MLP neurons, out of 12,288, whose firing rises most.
- **Seed agreement.** The Jaccard overlap of the set between seeds; 1 is identical.
- **Restore share.** Put the set alone back to the unchanged model's values; the share of the gain that disappears.
- **Re-switching share of the set.** Give the unchanged model only the set's change; the share of the gain recovered.

Each set reading sits beside 20 random same-size sets, the floor, and beside the full-width reading: VPD's $r$ from
E46 and E53, and E98's neuron shares $r_n$ and $r_\text{half}$. A panel "separates" by E12's filed rule: the classes'
ranges do not overlap, the gap exceeds the larger within-class range, and the existing class is higher.


## What we expected, and why

Guesses, with no pass or fail for the project. VPD's set would separate the maps on seed agreement, since E12 had read
Jaccards of about 0.8 against 0.6. It would not separate them on restore share, as E50 found. Neurons would separate
nothing, after E98 and E52 leaned the wrong way on the maps.


## Does the set repeat across seeds, for either method?

**Reading it.** VPD's set does, and the neurons' does not. VPD's copy tasks read 0.81 and 0.85, its random maps 0.55
and 0.66: a gap of 0.15 against a within-class spread of 0.11, so VPD separates. The neuron sets' gap is 0.05 against
a spread of 0.21. Seed agreement reads only whether the set recurs, not what it does.

## Does the set carry the gain?

**Reading it.** No, for either method, and neither separates the maps on it. Put back alone, VPD's set removes 0.14
to 0.23 of the gain on copy maps and 0.09 to 0.21 on random maps. Given alone, it recovers at most 0.02, where the full
width recovers up to 0.80. So VPD's re-switching does not live in the 0.3-threshold set. The neuron set recovers up to
0.14 but spreads as widely. VPD's re-switching set sits inside the range of its 20 random same-count sets on 4 of 8
states; the neuron set is above the random maximum on 7 of 8.

## At full width, which method orders the tasks as built?

**Reading it.** VPD does on both families; the neurons on one. These are E117's controls, read here by E12's rule as
description. VPD's $r$ puts copy maps 0.47 above random maps and lowercase 0.59 above the ciphers, both over E53's bar

## What we learned

On the 67M, VPD's two positive readings hold up against neurons: its set recurs across seeds by class, and its
full-width share orders both families. The set at the 0.3 threshold carries little gain for either method, and on
next-token tasks the window-mean rule leaves it empty. So what VPD adds here is in which subcomponents recur and in the full-width share, not in
a small set that carries the change.

## How compelling is the evidence

Two seeds per task; the classes are assumed from task design. Seed agreement is blind to what the set does; restore
share reads one position; the VPD re-switching keeps Δ at the unchanged model's. The W0 control read exactly 0, the
switched-on counts matched 41, 52, 27 and 21, and the full-width limits reproduced E46, E53 and E98.

## Surprises

No window-mean importance change passes 0.3 on lowercase or the ciphers at the read position. The plan expected a
VPD set there. Spread thin or averaged away over windows is open.

## What next

A lower importance threshold on the next-token tasks would be a new reading with its own plan; if it gave a set, the
lowercase panels would become readable.
