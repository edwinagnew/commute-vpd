# E110: Does the lowercase gain pass through the located pair, or only need it?

**Answer.** The two methods agree. The lowercase fine-tune needs heads (1,1) and (1,2): switching them off removes 0.80 and 0.85 of its gain. But neither their weights nor their output changed. The gain is carried by the MLPs of blocks 2 and 3. The pair is an existing part reused unchanged, and the fine-tune's change comes after it. No run was needed: E100 had already saved these reads (`experiments/fed_or_written/results/stage2.json`).

**The reads**, from E100's report. *Removal* is the gain lost when a part is set to the original model's mean output (E97). *Carry* is the gain lost when a part's output is replaced by the original's output on the same input. *Own* is the gain lost when the part's weight update alone is undone.

| Share of the lowercase gain, seeds 0 and 1 | Pair (1,1)+(1,2) | Block 3's MLP | Copy fine-tune at head (2,4), seeds 0-2 |
|---|---|---|---|
| Removal | 0.80, 0.85 | 1.93, 2.06 | 0.56-0.65 |
| Carry | 0.007, 0.012 | 0.62, 0.67 | 0.33-0.38 |
| Own | -0.005, -0.007 | 0.16, 0.16 | 0.005-0.008 |

The pair's carry sits under E100's floor of 0.021, the largest share any part inherits downstream of a planted edit. Copy is a different kind of reuse. Head (2,4)'s weights barely moved either, but its output did, and that change carries a third of the gain.

**What it means.** Removal finds parts a fine-tuned model needs; it cannot tell a part the gain runs through from one the gain only depends on. Carry can. Since 2026-11-06 the reuse definition is Jain et al.'s readout test (`decisions.md` rows 81, 87, 164), so this reading sits beside the definition and is not part of it.
