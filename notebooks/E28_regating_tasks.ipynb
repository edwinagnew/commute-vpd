# Does one fixed change of importance values carry the switching on new tasks? (E28)

*2026-10-22. Record, full tables and scoring: `reports/2026-10-22_analysis_regating-tasks.md`. Plan:
`reports/2026-10-22_plan_regating-tasks.md`. The hypothesis tested: `reports/2026-10-22_decision_working-hypothesis-and-two-lines.md`.*

## The answer

Partly. On the four new word tasks every test the working hypothesis could fail was passed. One fixed change of Γ's
importance values, the same for every input, gave back three quarters or more of what each input's own values give.
It worked only on the subcomponents it was aimed at, and another input's values did as well as the input's own. On the
three key-value tasks the fixed change gave back less than half. We had leaned the other way on the copy task, where
borrowing another input's values also failed.

## What we were trying to test

A ten-step fine-tune changes the weights by δW. We leave δW in Δ and keep the released subcomponents. Γ, the function
that sets each subcomponent's causal importance, is frozen. It still reads the updated model's activations, so it switches
some existing subcomponents up or down at the trigger. The *switching effect* is the rise in the CI-masked model's
target log-odds that this alone produces, with Δ left out.

The user's working hypothesis says that one fixed change to Γ's values at the trigger carries most of that effect. The
change is the same for every input and aimed at particular subcomponents. The decision report named three results on
a new task that would kill it, and this run reads all three:

- **Fixed change.** Give every input the updated model's mean values over the training windows. Subtract what the
  original model's mean values give, since averaging alone gives something back. Kill if under 0.5.
- **Shuffled change.** Move the change's values to other alive subcomponents of the same matrix, twenty draws. This
  rules out that raising and lowering this many values helps, whichever subcomponents get them. Kill if over 0.2.
- **Another input's values.** Give each window the updated values computed on a different window, five reshuffles. Its
  reference is that other window's original values. Kill if under 0.5, scored only where the reference is at most 0.2.

Every number is a share of what each input's own updated values give back at the trigger. Zero means the original
model's values; one means as good as the input's own. Two controls: the fixed change of a fine-tune to a different
target, ` http`, and the change's values moved to random subcomponents.

The tasks are four word tasks the model already does, mined from Pile text at rungs 0 to 2 of the task ladder. Three
more are key-value tasks it lacks. In those, after a marker, a word is followed by itself, the *copy task*, or by a fixed random
word, the *32-pair* and *one-pair ciphers*. Emoticon ran first and alone as the control. It reproduced E18 to the third
decimal on every number, including the shuffled and another-window arms. Each new task's switching share matched its
earlier run within 0.001.

## What we expected, and why

- **Word tasks: all three tests survive**, with medium-high confidence where switching carries most of the gain. E18's
  three tasks had passed.
- **Copy task: a lean to "survives", low confidence.** Switching carries 0.80 of its gain, as on the word tasks.
- **The two ciphers: no lean.** A fixed change can carry generic information. But another window has a different key,
  so failing that test would read "the switching carries the key".
- **Controls:** the wrong-target change and the random set give back at most 0.1.

## Does one fixed change give back at least half of what each input's own values give?

![fixed](../experiments/regating_tasks/figures/fixed.png)

**Reading it.** Yes on the four word tasks, no on the three key-value tasks. Each bar is the updated model's mean
values, net of the original model's mean values, as a share of the input's own values. On the word tasks the bars
read 0.74 to 0.99, inside or above the predicted band and well clear of the kill line. On the key-value tasks they
read 0.07 on the copy task and 0.15 on the 32-pair cipher. The one-pair cipher reads 0.45, just under the line. The
crosses are the subtracted reference, and they explain that narrow miss. There the updated mean gives back 0.90, but
the original model's mean already gives 0.45. On the copy task even the updated mean gives back only 0.33.

The open circles add the change to each input's own original values, the pattern E18 filed. On two word tasks they
read 0.40 and 0.45, where the mean pattern reads 0.74 and 0.88. E18 saw the same gap on its first word task. Its first
amendment traced it to the original model's input-to-input variation plus the clip; it was not re-measured here.

## Does the change stop working when its values go to the wrong subcomponents?

![shuffled](../experiments/regating_tasks/figures/shuffled.png)

**Reading it.** Yes on every task, but the test only has power on five of the seven. Shuffled, the change gives back
within 0.01 of zero on every task, against a kill line at 0.2. The open circles are the same change unshuffled. On
the word tasks and the one-pair cipher it gives back 0.40 to 0.82, so shuffling destroys a real effect. On the copy
task and the 32-pair cipher the unshuffled change gives back under 0.05. There is nothing for shuffling to destroy,
and survival there says nothing.

## Do another input's updated values work as well as the input's own?

![other](../experiments/regating_tasks/figures/other.png)

**Reading it.** Yes on the word tasks, no on the copy task, unreadable on the two ciphers. Each bar gives every window
the updated values computed on another window. The open circles are the reference: the original values of that other
window. On the word tasks another window's values give back 0.96 to 1.00, against references of 0.03 to 0.14. On the
copy task they give back 0.26 against a reference of 0.17. So on the copy task the updated values at the key carry
something specific to the input. The natural reading is the key's identity, which the copy task must reproduce; that
is a reading, not a measurement. On the two ciphers the reference itself is 0.30 and 0.28, above the 0.2 the rule
allows, so their bars are reported, not scored. The one-pair cipher has a single key, so this test could not read key
identity there.

The controls separate patterns on all eight states. The wrong-target change gives back −0.24 to +0.06. The random
set gives back nothing on average.

## What we learned

The working hypothesis survived every test it could fail on four tasks it had never been read on. With E18's three, it
now holds on seven tasks the model already does: emoticon and six word tasks across rungs 0 to 2. On the three tasks
the model lacks, a single fixed change does not carry the switching. On the two ciphers that lines up with E12. There
switching carries only 0.12 and 0.27 of the gain; the rest runs through subcomponents Γ keeps off. The copy task is the
new case. Switching carries 0.80 of its gain, as on the word tasks. Yet a fixed change gives back almost none of it,
and another input's values fail too. Whether the hypothesis is restated as "on tasks the model already does" is the
user's decision.

## How compelling is the evidence

Strong for the word tasks, narrower than it looks for the key-value tasks.

- One fine-tune seed per task, one 67M model, the trigger position only, about a hundred held-out windows per state.
- The readings are deterministic given the state; the control matched E18 within 0.002.
- The shuffle test has no power on the copy task and the 32-pair cipher.
- The fixed change is one per task, averaged over 32 keys on the 32-key tasks. A change per key was not tried.
- On the key-value tasks the original model's mean values already give back 0.26 to 0.45. On the word tasks they give
  0.09 to 0.28.
- The one-pair cipher's switching carries only 0.12 of its gain. The 32-pair cipher's fine-tune reached a held-out
  target probability of only 0.59.

## Surprises

- **A filed lean reversed on the copy task.** We leaned "survives" at low confidence, because switching carries 0.80
  of its gain. It failed two of the three conditions. The fixed change gave back 0.07, against a kill line at 0.5.
  Another input's values gave back 0.26 against a reference of 0.17. What the input-specific part is, we cannot say
  from these arms; the key's identity is the natural reading, not a measurement.
- **The filed pattern sat far below the mean pattern on two new word tasks.** It read 0.40 and 0.45 there. The net
  mean pattern read 0.74 and 0.88. E18 saw the same gap and traced it to the original model's input-to-input
  variation plus the clip.
- **Not predicted: the original model's mean values give back more on the key-value tasks.** On the one-pair cipher
  the input's own values even give back more than the whole switching effect.

## What next

1. **A fixed change per key on the copy task and the 32-pair cipher.** If it gives back 0.7 or more, the hypothesis
   holds per key. If it stays under 0.5, the switching there depends on the input beyond its key.
2. **The user's call on scope.** If the hypothesis is restated as "on tasks the model already does", E29 and the larger
   model go ahead.
